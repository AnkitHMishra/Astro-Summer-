# ⭐ Section 5: Stellar Astrophysics
## *The Physics of Stars — From Hydrogen to Supernovae*
---

### What is a Star?

A star is a **gravitationally bound plasma sphere** in which nuclear fusion reactions in the core release energy that supports the star against gravitational collapse. The Sun fuses ~$6 \times 10^{11}$ kg of hydrogen **every second**.

### 5.1 Hydrostatic Equilibrium

A star exists in a delicate balance between gravity (pulling inward) and radiation pressure + gas pressure (pushing outward):

$$\boxed{\frac{dP}{dr} = -\frac{G M(r) \rho(r)}{r^2}}$$

Where:
- $P$ = pressure at radius $r$
- $G = 6.674 \times 10^{-11}$ N m² kg⁻² (gravitational constant)
- $M(r)$ = mass enclosed within radius $r$
- $\rho(r)$ = density at radius $r$

### 5.2 The PP-Chain: How Stars Shine

In stars like the Sun, the dominant energy source is the **proton-proton chain**:

$$\text{Net reaction:} \quad 4\, {}^1\text{H} \rightarrow {}^4\text{He} + 2e^+ + 2\nu_e + 2\gamma$$

The energy released follows Einstein's famous relation:

$$\boxed{E = \Delta m \, c^2}$$

The mass defect per reaction: $\Delta m = 4 \times 1.007825 - 4.002602 = 0.028698$ u $= 4.75 \times 10^{-29}$ kg

$$E = 4.75 \times 10^{-29} \times (3 \times 10^8)^2 \approx 4.3 \times 10^{-12} \text{ J per reaction}$$

### 5.3 The Stefan-Boltzmann Law — Stellar Luminosity

A star's luminosity (total power output) depends on its surface temperature and radius:

$$\boxed{L = 4\pi R^2 \sigma T_{\text{eff}}^4}$$

Where $\sigma = 5.67 \times 10^{-8}$ W m⁻² K⁻⁴ is the Stefan-Boltzmann constant.

### 5.4 Wien's Displacement Law — Stellar Colors

The peak wavelength of a star's spectrum tells us its temperature:

$$\boxed{\lambda_{\text{max}} = \frac{b}{T}, \quad b = 2.898 \times 10^{-3} \text{ m K}}$$

For the Sun ($T \approx 5778$ K): $\lambda_{\text{max}} \approx 502$ nm — **peak in green/yellow light!**

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch

# ============================================================
# PLOT 1: Hertzsprung-Russell (H-R) Diagram (simulated)
# ============================================================
np.random.seed(42)

# Simulate stellar populations
# Main sequence: log L ≈ 4 * log(T/T_sun) roughly
T_sun = 5778

# Main sequence
T_ms = np.random.uniform(3000, 50000, 600)
L_ms = 10 ** (4.0 * np.log10(T_ms / T_sun) + np.random.normal(0, 0.15, 600))

# Giants (cooler, very luminous)
T_g = np.random.uniform(3500, 7000, 80)
L_g = 10 ** (np.random.uniform(2, 3.5, 80))

# Supergiants
T_sg = np.random.uniform(4000, 30000, 30)
L_sg = 10 ** (np.random.uniform(4, 6, 30))

# White dwarfs
T_wd = np.random.uniform(8000, 40000, 100)
L_wd = 10 ** (np.random.uniform(-3, -1, 100))

fig, axes = plt.subplots(1, 2, figsize=(18, 8))
fig.patch.set_facecolor('#0d1117')

# H-R Diagram
ax1 = axes[0]
ax1.set_facecolor('#0d1117')

# Color-code by temperature (B-V proxy)
def temp_to_color(T):
    T_norm = np.clip((T - 3000) / (40000 - 3000), 0, 1)
    return plt.cm.RdYlBu(T_norm)

ax1.scatter(T_ms, L_ms, c=T_ms, cmap='RdYlBu', s=8, alpha=0.7, norm=plt.Normalize(3000, 40000), label='Main Sequence')
ax1.scatter(T_g, L_g, c=T_g, cmap='RdYlBu', s=60, alpha=0.9, norm=plt.Normalize(3000, 40000), edgecolors='orange', linewidths=0.5, label='Giants')
ax1.scatter(T_sg, L_sg, c=T_sg, cmap='RdYlBu', s=120, alpha=0.9, norm=plt.Normalize(3000, 40000), edgecolors='yellow', linewidths=0.8, label='Supergiants')
ax1.scatter(T_wd, L_wd, color='lightblue', s=12, alpha=0.8, label='White Dwarfs')
ax1.scatter([T_sun], [1.0], color='yellow', s=200, zorder=10, edgecolors='white', linewidths=1.5, label='☀ Sun')

ax1.set_xscale('log')
ax1.set_yscale('log')
ax1.invert_xaxis()
ax1.set_xlabel('Effective Temperature (K)', color='white', fontsize=13)
ax1.set_ylabel(r'Luminosity ($L/L_\odot$)', color='white', fontsize=13)
ax1.set_title('Hertzsprung–Russell Diagram', color='white', fontsize=15, fontweight='bold')
ax1.tick_params(colors='white')
ax1.spines['bottom'].set_color('gray')
ax1.spines['left'].set_color('gray')
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)
leg = ax1.legend(facecolor='#1a1a2e', labelcolor='white', fontsize=9)

# Annotations
ax1.text(4500, 300, 'Red Giants', color='orange', fontsize=10, fontstyle='italic')
ax1.text(20000, 1e5, 'Supergiants', color='yellow', fontsize=10, fontstyle='italic')
ax1.text(20000, 5e-3, 'White Dwarfs', color='lightblue', fontsize=10, fontstyle='italic')
ax1.text(8000, 0.7, 'Main Sequence', color='white', fontsize=10, fontstyle='italic', rotation=-45)

# ============================================================
# PLOT 2: Blackbody Spectra for different stellar types
# ============================================================
ax2 = axes[1]
ax2.set_facecolor('#0d1117')

h = 6.626e-34
c_light = 3e8
k_B = 1.381e-23

def planck(lam, T):
    return (2*h*c_light**2 / lam**5) / (np.exp(h*c_light/(lam*k_B*T)) - 1)

wavelengths = np.linspace(100e-9, 2500e-9, 2000)
star_types = [
    (3500,  'M-type (Red Giant)',    '#ff4444'),
    (5778,  'G-type (Sun)',          '#ffdd44'),
    (10000, 'A-type (Sirius)',       '#aaddff'),
    (30000, 'O-type (Blue Giant)',   '#6699ff'),
]

for T, label, color in star_types:
    B = planck(wavelengths, T)
    B_norm = B / B.max()
    ax2.plot(wavelengths*1e9, B_norm, color=color, linewidth=2.5, label=f'{label}\n$T = {T}$ K')

# Visible light band
ax2.axvspan(380, 700, alpha=0.12, color='white', label='Visible light')
ax2.axvline(502, color='yellow', linestyle='--', alpha=0.5, linewidth=1)
ax2.text(510, 1.02, "Sun's peak", color='yellow', fontsize=8)

ax2.set_xlabel('Wavelength (nm)', color='white', fontsize=13)
ax2.set_ylabel('Normalised Spectral Radiance', color='white', fontsize=13)
ax2.set_title("Blackbody Spectra: Wien's Law", color='white', fontsize=15, fontweight='bold')
ax2.set_xlim(100, 2500)
ax2.tick_params(colors='white')
ax2.spines['bottom'].set_color('gray')
ax2.spines['left'].set_color('gray')
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
leg2 = ax2.legend(facecolor='#1a1a2e', labelcolor='white', fontsize=9, loc='upper right')

plt.tight_layout()
plt.savefig('stellar_physics.png', dpi=120, bbox_inches='tight', facecolor='#0d1117')
plt.show()
````

### 5.5 Stellar Evolution: The Life Cycle

A star's fate is entirely determined by its **initial mass**:

$$\text{Low mass} (M < 8M_\odot): \quad \text{Main Sequence} \rightarrow \text{Red Giant} \rightarrow \text{Planetary Nebula} \rightarrow \text{White Dwarf}$$

$$\text{High mass} (M > 8M_\odot): \quad \text{Main Sequence} \rightarrow \text{Red Supergiant} \rightarrow \text{Core-collapse Supernova} \rightarrow \text{Neutron Star / Black Hole}$$

The main-sequence lifetime scales approximately as:

$$\boxed{\tau_{\text{MS}} \approx \frac{E_{\text{fuel}}}{L} = \frac{\epsilon \cdot f \cdot M c^2}{L} \propto \frac{M}{L} \propto M^{1-\alpha}}$$

Using the mass-luminosity relation $L \propto M^4$:

$$\tau_{\text{MS}} \propto M^{-3}, \quad \tau_\odot \approx 10 \text{ Gyr}$$

So a star **10 times** as massive as the Sun lives only **$10^{-3} \times$** as long — about **10 million years**!

> 💡 **Connection to Daily Life:** Every atom of carbon, oxygen, nitrogen, and iron in your body was forged in a stellar core and dispersed by supernovae. You are, quite literally, **stardust**.

# 🌀 Section 6: Galactic Astrophysics
## *The Architecture of the Universe*
---

A galaxy is a gravitationally bound system of stars, stellar remnants, gas, dust, and **dark matter**. The Milky Way contains ~$2 \times 10^{11}$ stars and spans ~$100,000$ light-years.

### 6.1 Galactic Rotation Curves & Dark Matter

If only visible matter existed, the orbital velocity of stars should follow Keplerian rotation:

$$v_{\text{Keplerian}}(r) = \sqrt{\frac{GM(r)}{r}} \quad \xrightarrow{r > R_{\text{disk}}} \quad v \propto r^{-1/2}$$

**Observation** (Vera Rubin, 1970s): Rotation curves are **FLAT** at large radii:

$$v(r) \approx \text{const} \quad \Rightarrow \quad M(r) \propto r$$

This implies an **invisible halo** with density profile:

$$\boxed{\rho_{\text{DM}}(r) = \frac{\rho_0}{\left(\frac{r}{r_s}\right)\left(1 + \frac{r}{r_s}\right)^2}} \quad \text{(NFW Profile)}$$

Where $r_s$ is the scale radius and $\rho_0$ is a characteristic density.

**Dark Matter constitutes ~27% of the universe's total energy content.** We do not know what it is.

### 6.2 The Virial Theorem — Galaxy Clusters

For a gravitationally bound system in equilibrium:

$$\boxed{2\langle K \rangle + \langle U \rangle = 0}$$

For a galaxy cluster: $M_{\text{total}} = \frac{5 \sigma_v^2 R}{G}$

Fritz Zwicky used this in 1933 on the Coma Cluster and found the **mass was 400× more than visible** — the **first evidence for dark matter**.

````python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Galactic Rotation Curve: Observed vs Keplerian
# ============================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.patch.set_facecolor('#0d1117')

r = np.linspace(1, 30, 300)  # in kpc

# Bulge contribution (solid-body then Keplerian)
r_bulge = 3.0
v_bulge = np.where(r < r_bulge, 220 * r / r_bulge, 220 * np.sqrt(r_bulge / r))

# Disk contribution
v_disk_peak = 180
v_disk = v_disk_peak * (r / 5) * np.exp(1 - r / 5)
v_disk = np.maximum(v_disk, 0)

# Dark matter halo: gives flat curve
v_halo = 100 * np.sqrt(np.log(1 + r/5) / (r/5) - 1/(1+r/5)) * (r / np.sqrt(r**2 + 2**2))
v_halo = np.abs(v_halo) * 1.5

# Total
v_total = np.sqrt(v_bulge**2 + v_disk**2 + v_halo**2)

# Pure Keplerian (no DM)
v_kepler = np.sqrt(v_bulge**2 + v_disk**2)

ax1 = axes[0]
ax1.set_facecolor('#0d1117')
ax1.plot(r, v_total, color='#00ff88', linewidth=3, label='Observed (flat)')
ax1.plot(r, v_kepler, color='#ff6644', linewidth=2, linestyle='--', label='Without Dark Matter')
ax1.plot(r, v_bulge, color='#ffdd44', linewidth=1.5, linestyle=':', alpha=0.7, label='Bulge')
ax1.plot(r, v_disk, color='#44aaff', linewidth=1.5, linestyle=':', alpha=0.7, label='Disk')
ax1.plot(r, v_halo, color='#cc44ff', linewidth=1.5, linestyle=':', alpha=0.7, label='Dark Matter Halo')

ax1.set_xlabel('Galactocentric Radius (kpc)', color='white', fontsize=12)
ax1.set_ylabel('Rotation Velocity (km/s)', color='white', fontsize=12)
ax1.set_title('Galactic Rotation Curve\n(Evidence for Dark Matter)', color='white', fontsize=13, fontweight='bold')
ax1.tick_params(colors='white')
ax1.set_xlim(0, 30)
ax1.set_ylim(0, 320)
for sp in ['bottom','left']: ax1.spines[sp].set_color('gray')
for sp in ['top','right']: ax1.spines[sp].set_visible(False)
ax1.legend(facecolor='#1a1a2e', labelcolor='white', fontsize=9)
ax1.fill_between(r, v_kepler, v_total, alpha=0.15, color='#cc44ff')
ax1.text(15, 240, 'Missing Mass\n= Dark Matter', color='#cc44ff', fontsize=10, ha='center')

# ============================================================
# Galaxy Morphology (Hubble Sequence schematic)
# ============================================================
ax2 = axes[1]
ax2.set_facecolor('#0d1117')

theta = np.linspace(0, 2*np.pi, 500)

# Elliptical galaxies E0-E7
for i, b_over_a in enumerate([1.0, 0.85, 0.7, 0.55]):
    x_pos = -6 + i * 1.2
    x = 0.5 * np.cos(theta) + x_pos
    y = 0.5 * b_over_a * np.sin(theta)
    ax2.fill(x, y + 1.5, alpha=0.6, color='#ffddaa')
    ax2.text(x_pos, 0.6, f'E{2*i}', color='#ffddaa', fontsize=7, ha='center')

# Spiral galaxies (schematic)
for i, arm_tightness in enumerate([0.5, 0.8, 1.1]):
    x_pos = 0 + i * 2.2
    # Bulge
    x_b = 0.3*np.cos(theta) + x_pos
    y_b = 0.15*np.sin(theta)
    ax2.fill(x_b, y_b - 1.5, alpha=0.8, color='#ffcc66')
    # Arms
    r_arm = np.linspace(0.3, 1.2, 200)
    phi = arm_tightness * r_arm * 4
    for sign in [1, -1]:
        ax2.plot(r_arm*np.cos(phi + sign*np.pi) + x_pos, r_arm*np.sin(phi + sign*np.pi) - 1.5, 
                 color='#4488ff', alpha=0.7, linewidth=1.5)
    ax2.text(x_pos, -3.0, f'Sb{i}', color='#4488ff', fontsize=7, ha='center')

ax2.text(-4, 2.5, 'Elliptical Galaxies', color='#ffddaa', fontsize=11, fontweight='bold', ha='center')
ax2.text(2, 0.5, 'Spiral Galaxies', color='#4488ff', fontsize=11, fontweight='bold', ha='center')
ax2.axhline(0, color='gray', linestyle='--', alpha=0.3)

ax2.set_xlim(-7, 6)
ax2.set_ylim(-4, 3.5)
ax2.set_title('Hubble Sequence: Galaxy Morphology', color='white', fontsize=13, fontweight='bold')
ax2.axis('off')

plt.tight_layout()
plt.savefig('galactic_astrophysics.png', dpi=120, bbox_inches='tight', facecolor='#0d1117')
plt.show()
````

# 🌐 Section 7: Cosmology
## *The Study of the Universe as a Whole*
---

Cosmology asks the deepest questions: **How did the universe begin? How does it evolve? What is its fate?**

### 7.1 The Friedmann Equations

The dynamics of an expanding, homogeneous, isotropic universe are governed by the Friedmann equations, derived from Einstein's field equations:

$$\boxed{\left(\frac{\dot{a}}{a}\right)^2 = \frac{8\pi G \rho}{3} - \frac{kc^2}{a^2} + \frac{\Lambda c^2}{3}}$$

$$\boxed{\frac{\ddot{a}}{a} = -\frac{4\pi G}{3}\left(\rho + \frac{3P}{c^2}\right) + \frac{\Lambda c^2}{3}}$$

Where:
- $a(t)$ = **scale factor** (relative size of the universe)
- $H(t) = \dot{a}/a$ = **Hubble parameter**
- $k$ = curvature ($-1, 0, +1$)
- $\Lambda$ = **cosmological constant** (dark energy)
- $\rho$ = energy density, $P$ = pressure

### 7.2 Hubble's Law — The Expanding Universe

Edwin Hubble observed in 1929 that galaxies are receding with velocity proportional to distance:

$$\boxed{v = H_0 \, d}$$

Where $H_0 \approx 67.4 \text{ km s}^{-1} \text{ Mpc}^{-1}$ (Planck 2018).

The recession velocity is detected as a **redshift**:

$$z = \frac{\lambda_{\text{obs}} - \lambda_{\text{emit}}}{\lambda_{\text{emit}}} = \frac{a_{\text{now}}}{a_{\text{emit}}} - 1$$

### 7.3 The $\Lambda$CDM Model

The **standard model of cosmology** describes the universe's energy budget:

$$\Omega_{\text{total}} = \Omega_m + \Omega_r + \Omega_\Lambda = 1 \quad (\text{flat universe})$$

| Component | Symbol | Fraction |
|---|---|---|
| Ordinary matter | $\Omega_b$ | ~5% |
| Dark matter | $\Omega_{\text{DM}}$ | ~27% |
| Dark energy | $\Omega_\Lambda$ | ~68% |
| Radiation | $\Omega_r$ | ~0.005% |

### 7.4 The Cosmic Microwave Background (CMB)

~380,000 years after the Big Bang, the universe cooled enough for electrons and protons to combine into neutral hydrogen — the **epoch of recombination**. Photons could travel freely, creating the CMB.

Today, CMB temperature: $T_{\text{CMB}} = 2.725 \pm 0.001$ K

CMB anisotropies reveal density fluctuations described by the **angular power spectrum** $C_\ell$, encoding the entire history of the early universe.

# ⚡ Section 8: High-Energy Astrophysics
## *Black Holes, Neutron Stars & Gravitational Waves*
---

### 8.1 Black Holes — The Schwarzschild Radius

A **black hole** forms when mass is compressed within its **Schwarzschild radius** — the event horizon:

$$\boxed{r_s = \frac{2GM}{c^2}}$$

For the Sun: $r_s \approx 3$ km. For Earth: $r_s \approx 9$ mm.

For the supermassive black hole at the Milky Way centre (Sgr A*, $M \approx 4 \times 10^6 M_\odot$):

$$r_s = \frac{2 \times 6.674 \times 10^{-11} \times 4 \times 10^6 \times 2 \times 10^{30}}{(3 \times 10^8)^2} \approx 1.2 \times 10^{10} \text{ m} \approx 0.08 \text{ AU}$$

### 8.2 General Relativistic Time Dilation Near a Black Hole

$$\boxed{d\tau = dt\sqrt{1 - \frac{r_s}{r}}}$$

As $r \rightarrow r_s$, $d\tau \rightarrow 0$ — **time stops at the event horizon** for a distant observer.

### 8.3 Hawking Radiation — Black Holes Aren't Forever

Quantum mechanics predicts black holes slowly evaporate via **Hawking radiation** at temperature:

$$\boxed{T_H = \frac{\hbar c^3}{8\pi G M k_B}}$$

For a stellar black hole ($M \sim 10 M_\odot$): $T_H \sim 10^{-8}$ K — essentially undetectable.

### 8.4 Gravitational Waves — Ripples in Spacetime

Einstein's general relativity predicts that accelerating masses radiate gravitational waves. The strain $h$ detectable at distance $r$ from a binary merger:

$$\boxed{h \sim \frac{G}{c^4} \cdot \frac{\ddot{Q}}{r}}$$

Where $Q$ is the **mass quadrupole moment**.

LIGO detected the first gravitational wave on **14 September 2015** (GW150914) — two black holes of 29 and 36 $M_\odot$ merging 1.3 billion light-years away, radiating $3 M_\odot c^2$ in gravitational waves in **0.2 seconds** — briefly outshining the entire observable universe in gravitational luminosity.

> 💡 **Connection to Daily Life:** GPS satellites need relativistic corrections from both special and general relativity to maintain centimetre accuracy. Without these, GPS would drift by **~10 km per day**.

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

fig, axes = plt.subplots(1, 3, figsize=(20, 6))
fig.patch.set_facecolor('#0d1117')

# ============================================================
# Plot 1: Spacetime curvature near a black hole (embedding diagram)
# ============================================================
ax1 = axes[0]
ax1.set_facecolor('#0d1117')

r = np.linspace(0.1, 4, 300)
theta = np.linspace(0, 2*np.pi, 300)
R, Theta = np.meshgrid(np.linspace(-4, 4, 200), np.linspace(-4, 4, 200))
dist = np.sqrt(R**2 + Theta**2)
Z = -1.5 / (dist + 0.3)  # Funnel shape
Z[dist < 0.3] = -5  # Singularity

c = ax1.contourf(R, Theta, Z, levels=50, cmap='inferno')
# Schwarzschild radius circle
circle = Circle((0, 0), 0.5, color='black', zorder=10)
ax1.add_patch(circle)
circle2 = Circle((0, 0), 0.5, fill=False, edgecolor='#ff4400', linewidth=2.5, linestyle='--', zorder=11)
ax1.add_patch(circle2)
ax1.text(0, 0.75, 'Event Horizon\n$r_s = 2GM/c^2$', color='#ff8844', fontsize=9, ha='center')
ax1.text(2.5, 2.5, 'Spacetime\nCurvature', color='white', fontsize=9, ha='center')
ax1.set_title('Black Hole — Spacetime Warping', color='white', fontsize=12, fontweight='bold')
ax1.set_aspect('equal')
ax1.axis('off')

# ============================================================
# Plot 2: Simulated Gravitational Wave signal (GW150914-like)
# ============================================================
ax2 = axes[1]
ax2.set_facecolor('#0d1117')

t_gw = np.linspace(-0.4, 0.1, 5000)
# Chirp signal: frequency increases, then ringdown
tau = 0.1 - t_gw  # time to merger
tau_clipped = np.maximum(tau, 0.001)
f_chirp = 35 * (0.1 / tau_clipped)**0.375
f_chirp = np.clip(f_chirp, 0, 300)
amplitude_chirp = 0.5 * (0.05 / tau_clipped)**0.5
amplitude_chirp = np.clip(amplitude_chirp, 0, 5)

phase = np.cumsum(2 * np.pi * f_chirp * (t_gw[1] - t_gw[0]))
h = amplitude_chirp * np.sin(phase)

# Ringdown after merger
mask_after = t_gw > 0
h[mask_after] *= np.exp(-t_gw[mask_after] / 0.02)

h_scaled = h * 5e-22 / h.max()

ax2.plot(t_gw, h_scaled * 1e21, color='#00ff88', linewidth=1.5)
ax2.axvline(0, color='#ff4400', linestyle='--', alpha=0.8, linewidth=1.5, label='Merger')
ax2.fill_between(t_gw, h_scaled * 1e21, alpha=0.2, color='#00ff88')
ax2.set_xlabel('Time (s)', color='white', fontsize=12)
ax2.set_ylabel(r'Strain $h$ ($\times 10^{-21}$)', color='white', fontsize=12)
ax2.set_title('Gravitational Wave Signal\n(GW150914-like, simulated)', color='white', fontsize=12, fontweight='bold')
ax2.tick_params(colors='white')
for sp in ['bottom','left']: ax2.spines[sp].set_color('gray')
for sp in ['top','right']: ax2.spines[sp].set_visible(False)
ax2.legend(facecolor='#1a1a2e', labelcolor='white')

# ============================================================
# Plot 3: Pulsar timing — neutron star light curve
# ============================================================
ax3 = axes[2]
ax3.set_facecolor('#0d1117')

t_psr = np.linspace(0, 10, 10000)
P = 0.714  # Crab pulsar period in seconds (simplified)
# Gaussian pulses
pulse = np.zeros_like(t_psr)
for t0 in np.arange(0, 10, P):
    pulse += 2.5 * np.exp(-0.5*((t_psr - t0)/0.03)**2)
    pulse += 0.8 * np.exp(-0.5*((t_psr - t0 - 0.35)/0.06)**2)  # interpulse

pulse += np.random.normal(0, 0.1, len(t_psr))

ax3.plot(t_psr, pulse, color='#44aaff', linewidth=0.8, alpha=0.9)
ax3.set_xlabel('Time (s)', color='white', fontsize=12)
ax3.set_ylabel('Relative Flux', color='white', fontsize=12)
ax3.set_title(f'Pulsar Light Curve (P ≈ {P} s)\n"Cosmic Clocks" — More Accurate than Atomic Clocks', 
              color='white', fontsize=11, fontweight='bold')
ax3.tick_params(colors='white')
ax3.set_xlim(0, 5)
for sp in ['bottom','left']: ax3.spines[sp].set_color('gray')
for sp in ['top','right']: ax3.spines[sp].set_visible(False)

plt.tight_layout()
plt.savefig('high_energy_astro.png', dpi=120, bbox_inches='tight', facecolor='#0d1117')
plt.show()
````

# 🪐 Section 9: Planetary Science & Exoplanets
## *Are We Alone? The Search for Habitable Worlds*
---

### 9.1 Kepler's Laws — Orbital Mechanics

$$\boxed{T^2 = \frac{4\pi^2}{GM} a^3} \quad \text{(Kepler's Third Law)}$$

Where $T$ is the orbital period, $a$ is the semi-major axis, and $M$ is the central mass.

### 9.2 Transit Method — Detecting Exoplanets

When a planet passes in front of its star, it blocks a fraction of starlight:

$$\boxed{\frac{\Delta F}{F} = \left(\frac{R_p}{R_*}\right)^2}$$

Earth transiting the Sun: $\Delta F / F \approx (6371/696000)^2 \approx 8.4 \times 10^{-5}$ — 0.0084%!

NASA's **Kepler** mission used this method to discover >2,600 confirmed exoplanets.

### 9.3 The Habitable Zone — Where Life Might Exist

The habitable zone is where liquid water can exist on a planet's surface:

$$\boxed{r_{\text{HZ}} = \sqrt{\frac{L/L_\odot}{S_{\text{eff}}}} \text{ AU}}$$

Where $S_{\text{eff}}$ is the effective solar flux (ranging from ~0.97 to ~1.7 depending on model).

### 9.4 The Drake Equation — A Framework for Curiosity

Frank Drake (1961) estimated the number of detectable civilisations in our galaxy:

$$\boxed{N = R_* \cdot f_p \cdot n_e \cdot f_l \cdot f_i \cdot f_c \cdot L}$$

| Term | Meaning | Estimate |
|---|---|---|
| $R_*$ | Star formation rate | ~3/year |
| $f_p$ | Fraction with planets | ~1.0 |
| $n_e$ | Habitable planets per system | ~0.4 |
| $f_l$ | Fraction where life arises | **unknown** |
| $f_i$ | Fraction with intelligence | **unknown** |
| $f_c$ | Fraction that communicate | **unknown** |
| $L$ | Civilisation lifetime (years) | **unknown** |

> We have gone from no known exoplanets (pre-1995) to **5,700+ confirmed exoplanets** by 2024.

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse, Circle
import matplotlib.patches as mpatches

fig, axes = plt.subplots(1, 3, figsize=(20, 6))
fig.patch.set_facecolor('#0d1117')

# ============================================================
# Plot 1: Simulated transit light curve
# ============================================================
ax1 = axes[0]
ax1.set_facecolor('#0d1117')

t = np.linspace(-5, 5, 2000)
depth = 0.015  # ~Jupiter-sized
duration = 3.5  # hours
ingress = 0.5

flux = np.ones_like(t)
mask_full = np.abs(t) < (duration/2 - ingress)
mask_in = (t >= -(duration/2)) & (t <= -(duration/2 - ingress))
mask_out = (t >= (duration/2 - ingress)) & (t <= (duration/2))

flux[mask_full] = 1 - depth
flux[mask_in] = 1 - depth * (t[mask_in] + duration/2) / ingress
flux[mask_out] = 1 - depth * (duration/2 - t[mask_out]) / ingress

noise = np.random.normal(0, 0.002, len(t))
flux_obs = flux + noise

ax1.plot(t, flux_obs, color='#aaccff', linewidth=0.8, alpha=0.7, label='Observed')
ax1.plot(t, flux, color='#ff8844', linewidth=2.5, label='Model', zorder=5)
ax1.axhline(1.0, color='gray', linestyle='--', alpha=0.4)
ax1.annotate('', xy=(-(duration/2), 1-depth/2), xytext=(duration/2, 1-depth/2),
             arrowprops=dict(arrowstyle='<->', color='yellow', lw=1.5))
ax1.text(0, 1-depth/2 - 0.005, f'Transit Duration\n({duration:.1f} hours)', 
         color='yellow', fontsize=8, ha='center', va='top')
ax1.text(3.5, 1-depth+0.002, f'Depth = $(R_p/R_*)^2$\n$= {depth:.3f}$', color='#ff8844', fontsize=8)

ax1.set_xlabel('Time from Mid-Transit (hours)', color='white', fontsize=11)
ax1.set_ylabel('Relative Flux', color='white', fontsize=11)
ax1.set_title('Exoplanet Transit Light Curve', color='white', fontsize=13, fontweight='bold')
ax1.tick_params(colors='white')
for sp in ['bottom','left']: ax1.spines[sp].set_color('gray')
for sp in ['top','right']: ax1.spines[sp].set_visible(False)
ax1.legend(facecolor='#1a1a2e', labelcolor='white', fontsize=9)
ax1.set_xlim(-5, 5)

# ============================================================
# Plot 2: Exoplanet mass-radius diagram
# ============================================================
ax2 = axes[1]
ax2.set_facecolor('#0d1117')

np.random.seed(12)
n_planets = 300
# Rocky: M < 10 Earth masses
M_rocky = np.random.exponential(3, 100)
M_rocky = M_rocky[M_rocky < 10]
R_rocky = M_rocky**(0.27) + np.random.normal(0, 0.1, len(M_rocky))

# Neptune-like: 10-100
M_nept = np.random.uniform(10, 100, 80)
R_nept = 1.5 * M_nept**(0.15) + np.random.normal(0, 0.3, 80)

# Gas giants: >100
M_giant = np.random.uniform(100, 5000, 60)
R_giant = np.random.uniform(8, 15, 60) + np.random.normal(0, 1, 60)

ax2.scatter(M_rocky, R_rocky, c='#ff8844', s=20, alpha=0.8, label='Rocky')
ax2.scatter(M_nept, R_nept, c='#44bbff', s=20, alpha=0.8, label='Neptune-like')
ax2.scatter(M_giant, R_giant, c='#cc88ff', s=20, alpha=0.8, label='Gas Giant')

# Solar system planets
ss = {'Earth': (1, 1, 'white'), 'Neptune': (17.1, 3.86, '#44bbff'), 
      'Jupiter': (317.8, 11.2, '#ff8844'), 'Saturn': (95.2, 9.45, '#ffdd44')}
for name, (m, r, col) in ss.items():
    ax2.scatter(m, r, c=col, s=120, zorder=10, edgecolors='white', linewidths=1)
    ax2.text(m*1.1, r, name, color='white', fontsize=7)

ax2.set_xscale('log')
ax2.set_yscale('log')
ax2.set_xlabel(r'Planet Mass ($M_\oplus$)', color='white', fontsize=11)
ax2.set_ylabel(r'Planet Radius ($R_\oplus$)', color='white', fontsize=11)
ax2.set_title('Exoplanet Mass–Radius Diagram', color='white', fontsize=13, fontweight='bold')
ax2.tick_params(colors='white')
for sp in ['bottom','left']: ax2.spines[sp].set_color('gray')
for sp in ['top','right']: ax2.spines[sp].set_visible(False)
ax2.legend(facecolor='#1a1a2e', labelcolor='white', fontsize=9)

# ============================================================
# Plot 3: Habitable Zone diagram
# ============================================================
ax3 = axes[2]
ax3.set_facecolor('#0d1117')

L_values = np.logspace(-2, 4, 400)  # Solar luminosities
r_inner = np.sqrt(L_values / 1.76)
r_outer = np.sqrt(L_values / 0.32)

ax3.fill_betweenx(L_values, r_inner, r_outer, alpha=0.25, color='#44ff88', label='Habitable Zone')
ax3.plot(r_inner, L_values, color='#ff6644', linewidth=2)
ax3.plot(r_outer, L_values, color='#4488ff', linewidth=2)

planets_hz = [
    ('Earth', 1.0, 1.0, '#44aaff'),
    ('Mars', 1.52, 1.0, '#ff8844'),
    ('Venus', 0.72, 1.0, '#ffdd88'),
    ('Kepler-22b', 1.1, 0.79, '#88ff88'),
    ('Proxima b', 0.048, 0.0017, '#ff88ff'),
]
for name, a, L, col in planets_hz:
    ax3.scatter(a, L, c=col, s=80, zorder=10, edgecolors='white', linewidths=1)
    ax3.text(a * 1.15, L, name, color='white', fontsize=8)

ax3.set_xscale('log')
ax3.set_yscale('log')
ax3.set_xlabel('Orbital Radius (AU)', color='white', fontsize=11)
ax3.set_ylabel(r'Stellar Luminosity ($L/L_\odot$)', color='white', fontsize=11)
ax3.set_title('Habitable Zone: Where Liquid\nWater Can Exist', color='white', fontsize=13, fontweight='bold')
ax3.tick_params(colors='white')
for sp in ['bottom','left']: ax3.spines[sp].set_color('gray')
for sp in ['top','right']: ax3.spines[sp].set_visible(False)
ax3.legend(facecolor='#1a1a2e', labelcolor='white')

plt.tight_layout()
plt.savefig('exoplanets.png', dpi=120, bbox_inches='tight', facecolor='#0d1117')
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

# ============================================================
# Simulating a Gaia-like Colour-Magnitude Diagram (CMD)
# Using stellar evolutionary tracks
# ============================================================

np.random.seed(2024)

def generate_cmd_population(n_ms=800, n_rgb=120, n_wd=100, n_sg=20):
    """Generate a realistic CMD for an old open cluster (like M67)"""
    
    # Main sequence (B-V from -0.3 to 1.6, M_V from 10 to -2)
    bv_ms = np.random.uniform(-0.3, 1.6, n_ms)
    mv_ms = 4.5 * bv_ms + 2 + np.random.normal(0, 0.2, n_ms)  # approx MS
    mv_ms += 2  # shift for distance
    
    # Subgiant branch + Red Giant Branch
    bv_rgb = np.random.uniform(0.6, 1.7, n_rgb)
    mv_rgb = -4.0 * (bv_rgb - 0.65) + 4.2 + np.random.normal(0, 0.15, n_rgb)
    mv_rgb -= 1.5 * np.exp(-(bv_rgb - 1.2)**2 / 0.1)  # tip of RGB

    # White dwarfs
    bv_wd = np.random.uniform(-0.3, 0.5, n_wd)
    mv_wd = 8.5 + 2*bv_wd + np.random.normal(0, 0.3, n_wd)

    # Horizontal branch
    bv_hb = np.random.uniform(0.1, 0.9, 30)
    mv_hb = np.random.normal(0.5, 0.15, 30)
    
    return (bv_ms, mv_ms), (bv_rgb, mv_rgb), (bv_wd, mv_wd), (bv_hb, mv_hb)

ms, rgb, wd, hb = generate_cmd_population()

fig, axes = plt.subplots(1, 2, figsize=(16, 8))
fig.patch.set_facecolor('#0d1117')

# CMD
ax1 = axes[0]
ax1.set_facecolor('#0d1117')

# Color stars by B-V
all_bv = np.concatenate([ms[0], rgb[0], wd[0], hb[0]])
all_mv = np.concatenate([ms[1], rgb[1], wd[1], hb[1]])

sc = ax1.scatter(ms[0], ms[1], c=ms[0], cmap='RdYlBu_r', s=8, alpha=0.8, 
                  vmin=-0.3, vmax=1.8, label='Main Sequence')
ax1.scatter(rgb[0], rgb[1], c=rgb[0], cmap='RdYlBu_r', s=25, alpha=0.9,
            vmin=-0.3, vmax=1.8, edgecolors='none', label='RGB / Subgiant')
ax1.scatter(wd[0], wd[1], color='lightcyan', s=12, alpha=0.7, label='White Dwarfs')
ax1.scatter(hb[0], hb[1], color='yellow', s=30, alpha=0.9, marker='*', label='Horiz. Branch')

# Annotate regions
ax1.text(1.3, -2.5, 'Red Giant\nBranch', color='#ffaa44', fontsize=9, fontstyle='italic')
ax1.text(-0.1, 8.5, 'Main\nSequence', color='white', fontsize=9, fontstyle='italic', rotation=-60)
ax1.text(0.1, 11.5, 'White\nDwarfs', color='lightcyan', fontsize=9, fontstyle='italic')
ax1.text(0.35, -0.3, 'Horizontal\nBranch', color='yellow', fontsize=9, fontstyle='italic')

# Sun position
ax1.scatter([0.65], [4.83], color='yellow', s=200, marker='$☀$', zorder=15)

ax1.invert_yaxis()
ax1.set_xlabel('Colour Index (B−V)', color='white', fontsize=13)
ax1.set_ylabel(r'Absolute Magnitude $M_V$', color='white', fontsize=13)
ax1.set_title('Colour-Magnitude Diagram\n(Gaia-like Simulated Data)', 
              color='white', fontsize=14, fontweight='bold')
ax1.tick_params(colors='white')
for sp in ['bottom','left']: ax1.spines[sp].set_color('gray')
for sp in ['top','right']: ax1.spines[sp].set_visible(False)
leg = ax1.legend(facecolor='#1a1a2e', labelcolor='white', fontsize=9)
plt.colorbar(sc, ax=ax1, label='B-V Color Index').ax.yaxis.set_tick_params(color='white')

# ============================================================
# Redshift survey — simulating an SDSS-like galaxy redshift histogram
# ============================================================
ax2 = axes[1]
ax2.set_facecolor('#0d1117')

# Simulate galaxy redshift distribution (mimicking SDSS main galaxy sample)
z_low = np.random.beta(2, 18, 5000) * 0.25  # peak around z~0.1
z_cluster1 = np.random.normal(0.08, 0.005, 300)  # cluster at z=0.08
z_cluster2 = np.random.normal(0.14, 0.006, 250)
z_cluster3 = np.random.normal(0.21, 0.007, 200)
z_all = np.concatenate([z_low, z_cluster1, z_cluster2, z_cluster3])
z_all = z_all[(z_all > 0) & (z_all < 0.3)]

n, bins, patches = ax2.hist(z_all, bins=120, color='#4488ff', alpha=0.8, edgecolor='none')

# Color bars by distance
for i, (patch, b) in enumerate(zip(patches, bins[:-1])):
    hue = b / 0.3
    patch.set_facecolor(plt.cm.plasma(0.2 + hue * 0.6))

# Mark clusters
for z_c, name in [(0.08, 'Cluster A'), (0.14, 'Cluster B'), (0.21, 'Cluster C')]:
    ax2.axvline(z_c, color='#ff8844', linestyle='--', alpha=0.8, linewidth=1.5)
    ax2.text(z_c + 0.003, n.max()*0.85, name, color='#ff8844', fontsize=9, rotation=90)

ax2.set_xlabel('Redshift z', color='white', fontsize=13)
ax2.set_ylabel('Number of Galaxies', color='white', fontsize=13)
ax2.set_title('Galaxy Redshift Survey\n(SDSS-like, simulated) — Large Scale Structure', 
              color='white', fontsize=14, fontweight='bold')
ax2.tick_params(colors='white')
for sp in ['bottom','left']: ax2.spines[sp].set_color('gray')
for sp in ['top','right']: ax2.spines[sp].set_visible(False)

# Distance axis
ax2_top = ax2.twiny()
ax2_top.set_xlim(ax2.get_xlim())
z_ticks = np.array([0.05, 0.10, 0.15, 0.20, 0.25])
d_ticks = (z_ticks / 0.067).astype(int) * 10  # approx Mpc (z≈d*H0/c)
ax2_top.set_xticks(z_ticks)
ax2_top.set_xticklabels([f'{d} Mpc' for d in d_ticks], color='#aaddff', fontsize=8)
ax2_top.tick_params(colors='#aaddff')
ax2_top.spines['top'].set_color('gray')

plt.tight_layout()
plt.savefig('real_data.png', dpi=120, bbox_inches='tight', facecolor='#0d1117')
plt.show()

print("\n📡 Data sources to explore:")
print("  • Gaia: https://gea.esac.esa.int/")
print("  • NASA Exoplanet Archive: https://exoplanetarchive.ipac.caltech.edu/")
print("  • SDSS SkyServer: https://skyserver.sdss.org/")
print("  • LIGO Open Data: https://gwosc.org/")
print("  • NASA ADS (papers): https://ui.adsabs.harvard.edu/")
````
